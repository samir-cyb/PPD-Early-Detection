```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('/content/drive/MyDrive/thesis_ppd/PPD_data.csv')
print(df.shape)
df.head(10)
```

```python
df.drop(columns=['sr'], inplace=True)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace("’", "'", regex=False)
)

df.rename(columns={
    "Recieved Support": "Received Support",
    "rouble concentrating on things": "Trouble concentrating on things",
    "Birth compliancy": "Birth complications"
}, inplace=True)

df.columns.tolist()
```

**FRequncies**

```python
import pandas as pd

# Identify categorical columns (object type)
categorical_cols = df.select_dtypes(include=['object']).columns

# Loop through each categorical column and display frequency/percentage
for col in categorical_cols:
    print(f"--- Frequency and Percentage for: {col} ---")

    counts = df[col].value_counts(dropna=False)
    percentages = df[col].value_counts(dropna=False, normalize=True) * 100

    # Combine into a summary table
    cat_stats = pd.DataFrame({
        'Frequency': counts,
        'Percentage (%)': percentages
    })

    display(cat_stats)
    print("\n")
```

```python
df['Education Level'] = df['Education Level'].replace({
    'High school': 'High School',
    'Primary school': 'Primary School'
})

# Verify the changes
print(df['Education Level'].value_counts())
```

```python
df["Husband's education level"] = df["Husband's education level"].replace({
    'High school': 'High School',
    'Primary school': 'Primary School'
})

# Verify the changes
print(df["Husband's education level"].value_counts())
```

Total children column fixed


```python
df['Total children'] = df['Total children'].replace('More than two', 'More than Two')

# Verify the change
print("Updated 'Total children' counts:")
print(df['Total children'].value_counts())
```

HIstory of pregnancy loss

```python
df['History of pregnancy loss'] = df['History of pregnancy loss'].replace('Still-born delivery', 'Still-born Delivery')

# Verify the change
print("Updated 'History of pregnancy loss' counts:")
print(df['History of pregnancy loss'].value_counts(dropna=False))
```

```python
col_target = 'Disease before pregnancy'
counts = df[col_target].value_counts(dropna=False)
percentages = df[col_target].value_counts(dropna=False, normalize=True) * 100

freq_table = pd.DataFrame({
    'Frequency': counts,
    'Percentage (%)': percentages
})

print(f'--- Frequency Distribution for: {col_target} ---')
display(freq_table)
```

```python
import numpy as np

# Target column name (stripping spaces to be safe)
col_target = 'Disease before pregnancy'

# Replace the specific '1.2' value with NaN
df[col_target] = df[col_target].replace('1.2', np.nan)

# Display the new frequencies including NaNs
counts = df[col_target].value_counts(dropna=False)
percentages = df[col_target].value_counts(dropna=False, normalize=True) * 100

freq_table = pd.DataFrame({
    'Frequency': counts,
    'Percentage (%)': percentages
})

print(f'--- Updated Frequencies for: {col_target} ---')
display(freq_table)
```

```python
#Fix inconsistent category values
# -----------------------------
df = df.replace({

    'Chronic disease': 'Chronic Disease',
    'Non-chronic disease': 'Non-Chronic Disease',
    'Non-chronic Disease': 'Non-Chronic Disease',
})
```

```python
col_target = 'Diseases during pregnancy'
counts = df[col_target].value_counts(dropna=False)
percentages = df[col_target].value_counts(dropna=False, normalize=True) * 100

freq_table = pd.DataFrame({
    'Frequency': counts,
    'Percentage (%)': percentages
})

print(f'--- Frequency Distribution for: {col_target} ---')
display(freq_table)
```

```python
# Clean column names
df.columns = df.columns.str.strip()

# Clean text values in all object/categorical columns
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()

# Replace inconsistent labels
replacement_map = {
    'মাঝে মধ্যে (Several days)': 'Several days',
    'Several Days': 'Several days',
    'several days': 'Several days'
}

df = df.replace(replacement_map)

# Verify
for col in df.select_dtypes(include='object').columns:
    if df[col].astype(str).str.contains('Several', case=False, na=False).any():
        print("\nColumn:", col)
        print(df[col].value_counts(dropna=False))
```

```python
cols_to_fix = ['Occupation before latest pregnancy', 'Occupation After Your Latest Childbirth']

# Correct spelling
for col in cols_to_fix:
    df[col] = df[col].replace('House wife', 'Housewife')

# Check frequencies
for col in cols_to_fix:
    print(f'--- Frequency Distribution for: {col} ---')
    counts = df[col].value_counts(dropna=False)
    percentages = df[col].value_counts(dropna=False, normalize=True) * 100

    summary = pd.DataFrame({
        'Frequency': counts,
        'Percentage (%)': percentages
    })
    display(summary)
    print('\n')
```

```python
# Check duplicates
duplicate_count = df.duplicated().sum()
print("Duplicate rows:", duplicate_count)

print("Shape after removing duplicates:", df.shape)
```
